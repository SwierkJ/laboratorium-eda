```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("../data/sprzedaz.csv", sep=";")
df.head()
```

```python
print("Wymiary (wiersze, kolumny):", df.shape)
print("\nTypy danych:\n", df.dtypes)
print("\nBraki danych:\n", df.isna().sum())
print("\nLiczba duplikatów:", df.duplicated().sum())
```

```python
print("Statystyki opisowe:")
display(df["wartosc"].describe())

print("\nZapis kanałów sprzedaży przed czyszczeniem:")
print(df["kanal"].value_counts())
```

```python
# Ujednolicenie: usunięcie spacji i zamiana na małe litery
df["kanal"] = df["kanal"].astype(str).str.strip().str.lower()

print("1. Liczba zamówień w każdym kanale:\n", df["kanal"].value_counts())
print(
    "\n2. Suma wartości zamówień w każdym kanale:\n",
    df.groupby("kanal")["wartosc"].sum(),
)
print("\n3. Średnia wartość zamówienia ogółem:", df["wartosc"].mean())
print("4. Mediana wartości zamówienia ogółem:", df["wartosc"].median())
```

```python
sprzedaz = df.groupby("kanal")["wartosc"].sum()

sprzedaz.plot(
    kind="bar",
    title="Suma sprzedaży według kanału",
    ylabel="Wartość [zł]",
    xlabel="Kanał sprzedaży",
    rot=0,
)
plt.tight_layout()
plt.show()
```

### Wnioski z analizy:
1. Pomimo identycznej liczby zamówień w obu kanałach (po 6 zamówień), kanał internetowy wygenerował niemal dwukrotnie wyższą sumę sprzedaży (1490 zł) niż sklep stacjonarny (760 zł).
2. Średnia wartość pojedynczego koszyka w kanale internetowym (~248,33 zł, mediana 235 zł) jest znacznie wyższa niż w sklepie stacjonarnym (~126,67 zł, mediana 125 zł), co oznacza, że klienci kupujący online wydają przeciętnie więcej na jedno zamówienie.
